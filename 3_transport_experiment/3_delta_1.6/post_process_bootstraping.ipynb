```python
import pickle
import numpy as np
#### bootstrapping the depolarizing rate #### 
n, time_steps = 40, 15
Even_depo = np.zeros(shape=(n, time_steps))
for i in range(100):
    with open('bootstrapped_even_depo/bootstrapped_even_depo_'+str(i+1), 'rb') as f: 
        even_depo = pickle.load(f)
    Even_depo += even_depo
Even_depo = Even_depo/100

Odd_depo = np.zeros(shape=(n, time_steps))
for i in range(100):
    with open('bootstrapped_odd_depo/bootstrapped_odd_depo_'+str(i+1), 'rb') as f: 
        odd_depo = pickle.load(f)
    Odd_depo += odd_depo
Odd_depo = Odd_depo/100
```

```python
# create the JJ matrix for each bootstrapped datat 
boot_sample = 100
time_steps = 15
odr_Cjj = np.zeros(shape =(boot_sample, time_steps))
odr_local_cjj = []
for index in range(100):    
    with open('bootstrapped_even/bootstrapped_even_'+str(index+1), 'rb') as f: 
        even = pickle.load(f)
    with open('bootstrapped_odd/bootstrapped_odd_'+str(index+1), 'rb') as f: 
        odd = pickle.load(f)

    depo_jj_matrix  = np.zeros(shape=(n, 15)) 
    start = n//2 -2 
    for time_step in range(15):
        for pair in range(time_step+2):
            depo_iz = even[time_step][pair][0] / Even_depo[start+2*pair+1 ,time_step]
            depo_zi = even[time_step][pair][1] / Even_depo[start+2*pair ,time_step]
            depo_jj_matrix[start+pair*2][time_step] = (depo_iz-depo_zi)*-2/16
        start -= 1

    start =  n//2 -1 
    for time_step in range(15):
        for pair in range(time_step+1): 
            depo_iz = odd[time_step][pair][0] / odd_depo[start+2*pair+1 ,time_step]
            depo_zi = odd[time_step][pair][1] / odd_depo[start+2*pair ,time_step]
            depo_jj_matrix[start+pair*2][time_step] = (depo_iz-depo_zi)*-2/16
        start -= 1
    odr_local_cjj.append(depo_jj_matrix)
    odr_Cjj[index,:] = np.sum(depo_jj_matrix, axis = 0)

with open('odr_Cjj_bootstrapped', 'wb') as f:
    pickle.dump(odr_Cjj, f )

with open('odr_local_Cjj_bootstrapped', 'wb') as f:
    pickle.dump(np.array(odr_local_cjj), f )
```

```python
# create the JJ matrix for each bootstrapped datat 
boot_sample = 100
time_steps = 15
raw_Cjj = np.zeros(shape =(boot_sample, time_steps))
raw_local_Cjj = []
for index in range(100):    
    with open('bootstrapped_even/bootstrapped_even_'+str(index+1), 'rb') as f: 
        even = pickle.load(f)
    with open('bootstrapped_odd/bootstrapped_odd_'+str(index+1), 'rb') as f: 
        odd = pickle.load(f)

    jj_matrix  = np.zeros(shape=(n, 15)) 
    start = n//2 -2 
    for time_step in range(15):
        for pair in range(time_step+2):
            iz = even[time_step][pair][0] 
            zi = even[time_step][pair][1]
            jj_matrix[start+pair*2][time_step] = (iz-zi)*-2/16
        start -= 1

    start =  n//2 -1 
    for time_step in range(15):
        for pair in range(time_step+1): 
            iz = odd[time_step][pair][0]
            zi = odd[time_step][pair][1] 
            jj_matrix[start+pair*2][time_step] = (iz-zi)*-2/16
        start -= 1
    raw_local_Cjj.append(jj_matrix)
    raw_Cjj[index,:] = np.sum(jj_matrix, axis = 0)

with open('raw_Cjj_bootstrapped', 'wb') as f:
    pickle.dump(raw_Cjj, f )

with open('raw_local_Cjj_bootstrapped', 'wb') as f:
    pickle.dump(np.array(raw_local_Cjj), f )
```
